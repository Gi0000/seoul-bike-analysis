```python
import time
```

```python
# 1000행 불러오는 시간
start_time = time.perf_counter()
pd.read_csv(rental_log_path, nrows = 1000, encoding = 'euc-kr')
end_time = time.perf_counter()

end_time - start_time
```

```python
# 10000행 불러오는 시간
start_time = time.perf_counter()
pd.read_csv(rental_log_path, nrows = 10000, encoding = 'euc-kr')
end_time = time.perf_counter()

end_time - start_time
```

```python
# 1000행 불러오는 시간 3회 반복
elapsed_times = []

print("[1,000행]")
for i in range(3):
    start_time = time.perf_counter()
    pd.read_csv(rental_log_path, nrows = 1000, encoding = 'euc-kr')
    end_time = time.perf_counter()
    elapsed_time = end_time - start_time
    elapsed_times.append(elapsed_time)
    print("{0}회: {1:.4f}초".format(i+1, elapsed_time))

elapsed_times_avg = sum(elapsed_times) / len(elapsed_times)
print(f"평균: {elapsed_times_avg:.4f}초")
```

```python
# 10000행 불러오는 시간 3회 반복
elapsed_times = []

print("[10,000행]")
for i in range(3):
    start_time = time.perf_counter()
    pd.read_csv(rental_log_path, nrows = 10000, encoding = 'euc-kr')
    end_time = time.perf_counter()
    elapsed_time = end_time - start_time
    elapsed_times.append(elapsed_time)
    print("{0}회: {1:.4f}초".format(i+1, elapsed_time))

elapsed_times_avg = sum(elapsed_times) / len(elapsed_times)
print(f"평균: {elapsed_times_avg:.4f}초")
```
